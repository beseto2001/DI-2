```python
## ✅ 문제 1. 추천 데이터 구성하기
#- MBTI별 추천 정보를 저장할 딕셔너리를 만들어봅시다.
#- 각 MBTI는 문자열(예: `"INFP"`, `"ESTJ"` 등)로 표현됩니다.
#- 각 MBTI에는 여러 카테고리(예: `"취미"`, `"음악"`, `"활동"`)가 있으며,
#- 각 카테고리에는 여러 개의 추천 항목이 포함됩니다.
#- 최소 2개 이상의 MBTI, 각 MBTI당 2개 이상의 카테고리를 포함해봅시다.

recommendations = {
    "INFP": {
        "취미": ["독서", "글쓰기", "예술 감상"],
        "직업": ["작가", "예술가", "상담사"],
        "성격": ["창의적", "감성적", "이상주의"],
    },
    "INFT": {   
        "취미": ["독서", "철학", "자기계발"],
        "직업": ["상담사", "연구자", "교사"],
        "성격": ["이해심", "책임감", "성찰적"],
    },
    "INTP": {
        "취미": ["논리적 사고", "문제 해결", "게임"],
        "직업": ["연구원", "프로그래머", "분석가"],
        "성격": ["분석적", "호기심", "독립적"],
    },
    "INTJ": {
        "취미": ["전략 게임", "독서", "계획 세우기"],
        "직업": ["전략가", "연구원", "경영자"],
        "성격": ["계획적", "분석적", "결단력"],
    },
    "ISFP": {
        "취미": ["음악 감상", "자연 탐험", "예술 활동"],
        "직업": ["디자이너", "예술가", "사진작가"],
        "성격": ["감성적", "자유로운", "관찰력"],
    },
    "ISFJ": {
        "취미": ["봉사활동", "요리", "정리정돈"],
        "직업": ["간호사", "교사", "사회복지사"],
        "성격": ["책임감", "배려심", "조용한"],
    },
}   

```

```python
### 전체 데이타 보기 좋게 출력
for mbti, categories in recommendations.items():
    print(f"MBTI: {mbti}")
    for cat, item in categories.items():
        print(f"  {cat}: {', '.join(item)}")
```

```python
## ✅ 문제 2. 사용자 입력 처리하기
#- 사용자로부터 이름과 MBTI를 입력받아 튜플 형태로 저장합니다.
#- MBTI는 반드시 대문자로 변환해야 합니다.

name = input("당신의 이름을 입력하세요: ").strip()
mbti = input("당신의 MBTI를 입력하세요 (예: INFP, ENFP 등): " ).strip().upper()

#튜플로 저장
user_info = (name, mbti)

## ✅ 문제 3. 추천 데이터 출력 함수 만들기
#- MBTI별 추천 정보를 출력해 봅시다.
#- `print_recommendation(user_info)` 함수를 만듭니다.
#    - 함수명과 변수명을 적당하게 바꿔도 좋습니다.
#- 해당 MBTI가 `recommendations`에 있다면 추천 항목을 출력하고, 없다면 안내 문구를 출력합니다.

def print_recommendation(user_info):
    name, mbti = user_info

    if mbti in recommendations:
        print(f"안녕하세요, {name}님! 당신의 MBTI는 {mbti}입니다.")
        for category, items in recommendations[mbti].items():
            print(f"{category} 추천:")
            for item in items:
                print(f"  - {item}")
    else:
        print("입력하신 MBTI가 유효하지 않습니다.")

print_recommendation(user_info)
```

```python
## ✅ 문제 4. 추천 항목 추가 함수

def add_recommendation(mbti, category, item):
    #mbti = input("MBTI를 입력하세요: ").upper()

    if mbti not in recommendations:          # MBTI가 없으면 새로 만들기
        recommendations[mbti] = {}

    if category not in recommendations[mbti]:  # 카테고리가 없으면 새로 만들기
        recommendations[mbti][category] = []

    if item in recommendations[mbti][category]:  # 중복 방지
        print(f"'{item}'은(는) 이미 {mbti}의 {category}에 있어요.")
        return

    recommendations[mbti][category].append(item)
    print(f"{mbti}의 {category}에 '{item}'을(를) 추가했어요.")
```

```python
## 추천 항목 추가하기
mbti = mbti.upper()
category = input("카테고리를 입력하세요 (예: 취미, 직업, 음악): ")
item = input("추가할 항목을 입력하세요: ")
print(f"\n{mbti}의 {category} 카테고리에 '{item}'을(를) 추가합니다.")
    
add_recommendation(mbti, category, item)
print_recommendation(user_info)  # 추가된 결과 다시 출력
```
